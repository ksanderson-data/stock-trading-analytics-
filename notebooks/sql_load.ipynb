# Yahoo Finance Table Load, 5 years of S&P 500 / SPY Historical Data #

```python
import pandas as pd
from sqlalchemy import create_engine

# Configuration
csv_file_path = r'C:\Users\Kyle\stock-trading-analytics-\data\spy_5y.csv' # Replace with your CSV file path
table_name = 'yahoo_finance'  # The name you want for your SQL table
database_connection = "mssql+pyodbc://@localhost/portfolio_analytics?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
  # Replace with your database connection

engine = create_engine(database_connection)
# For PostgreSQL: engine = create_engine('postgresql://username:password@localhost:5432/database_name')
# For MySQL: engine = create_engine('mysql+pymysql://username:password@localhost/database_name')

# Read the CSV file
print(f"Reading CSV file: {csv_file_path}")
df = pd.read_csv(csv_file_path)

# Optional: Display the first few rows to verify data
print("Preview of the data:")
print(df.head())

# Optional: Clean column names (replace spaces with underscores, remove special characters)
#df.columns = [col.lower().replace(' ', '_').replace('-', '_') for col in df.columns]

# Create table and load data
print(f"Creating table '{table_name}' and loading data...")
df.to_sql(
    name=table_name,
    con=engine,
    if_exists='replace',  # Options: 'fail', 'replace', or 'append'
    index=False,  # Set to True if you want to include the DataFrame index as a column
    chunksize=1000  # Rows will be inserted in chunks of this size
)

print(f"Successfully loaded {len(df)} rows into table '{table_name}'")
```
